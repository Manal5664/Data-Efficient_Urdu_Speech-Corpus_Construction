```python
from phonemizer import phonemize
import gc  # garbage collector
import os

input_file = "/content/drive/MyDrive/Thesis 2/urmono.txt"
output_file = "/content/drive/MyDrive/Thesis 2/ipa_output.txt"
progress_file = "/content/drive/MyDrive/Thesis 2/progress.txt"

batch_size = 10

# resume from last line
if os.path.exists(progress_file):
    try:
        with open(progress_file, "r") as f:
            content = f.read().strip()

            if content:
                start_line = int(content)
            else:
                start_line = 0
                print("progress file empty, start from 0")

    except Exception as e:
        start_line = 0
        print("progress file corrupt:", e)

else:
    start_line = 0

print("Resuming from line:", start_line)

batch = []

with open(input_file, "r", encoding="utf-8", errors="ignore") as f_in, \
     open(output_file, "a", encoding="utf-8") as f_out:

    for i, line in enumerate(f_in):

        if i < start_line:
            continue

        sentence = line.strip()

        if not sentence:
            continue

        batch.append(sentence)

        if len(batch) >= batch_size:
            try:
                ipa_list = phonemize(
                    batch,
                    language="ur",
                    backend="espeak",
                    strip=True89
                )

                f_out.write("\n".join(ipa_list) + "\n")

            except Exception as e:
                print(f"Error at line {i}: {e}")

            batch.clear()
            gc.collect()

            # save progress
            with open(progress_file, "w") as pf:
                pf.write(str(i))

            print(f"Checkpoint saved at line {i}")

print("IPA conversion completed")
```

```python
# ==========================================================
# STEP 1: Find multi-character IPA patterns
# ==========================================================

import re
from collections import Counter

input_file = "/content/drive/MyDrive/Thesis 2/ipa_output.txt"

pattern_counter = Counter()

# Common multi-character IPA patterns
pattern = re.compile(
    r"(aː|iː|uː|eː|oː|"
    r"bʰ|pʰ|tʰ|dʰ|kʰ|gʰ|"
    r"tʃ|dʒ|ʈʰ|ɖʰ|"
    r"t̪|d̪|"
    r"n̪|l̪|r̪)"
)

with open(input_file, "r", encoding="utf-8", errors="ignore") as f:

    for line in f:

        matches = pattern.findall(line)

        pattern_counter.update(matches)

print("="*50)
print("Detected Multi-character IPA Symbols")
print("="*50)

for symbol, freq in pattern_counter.most_common():
    print(f"{symbol:6} -> {freq}")
```

```python
# ==========================================================
# STEP 2: Save detected IPA symbols
# ==========================================================

output_file = "/content/drive/MyDrive/Thesis 2/ipa_symbols_found.txt"

with open(output_file, "w", encoding="utf-8") as f:

    for symbol, freq in pattern_counter.most_common():
        f.write(f"{symbol}\t{freq}\n")

print("Saved:", output_file)
```

```python
# ==========================================================
# STEP 3: IPA Merge Mapping
# ==========================================================

merge_map = {

    # Long vowels
    "aː": "A",
    "iː": "I",
    "uː": "U",
    "eː": "E",
    "oː": "O",

    # Aspirated consonants
    "bʰ": "B",
    "pʰ": "P",
    "tʰ": "T",
    "dʰ": "D",
    "kʰ": "K",
    "gʰ": "G",

    # Affricates
    "tʃ": "C",
    "dʒ": "J",

    # Retroflex aspirates
    "ʈʰ": "X",
    "ɖʰ": "Y",

    # Dental consonants
    "t̪": "q",
    "d̪": "w",

    # Optional
    "n̪": "N",
    "l̪": "L",
    "r̪": "R"
}

print("Total mappings:", len(merge_map))
```

```python
# ==========================================================
# STEP 4: Create merged IPA corpus
# ==========================================================

input_file = "/content/drive/MyDrive/Thesis 2/ipa_output.txt"

output_file = "/content/drive/MyDrive/Thesis 2/ipa_merged.txt"

count = 0

with open(input_file, "r", encoding="utf-8", errors="ignore") as fin, \
     open(output_file, "w", encoding="utf-8") as fout:

    for line in fin:

        for old, new in merge_map.items():
            line = line.replace(old, new)

        fout.write(line)

        count += 1

        if count % 100000 == 0:
            print(f"Processed {count} lines")

print("="*50)
print("Merged corpus created successfully")
print("Saved at:", output_file)
print("="*50)
```

```python
# ==========================================================
# PART 2
# Character-level Scoring Function
# ==========================================================

from collections import Counter

# ==========================================================
# CURRENT COVERAGE
# Initially empty
# ==========================================================

covered_unigrams = Counter()

covered_bigrams = Counter()

# ==========================================================
# WEIGHTS
# ==========================================================

ALPHA = 1.0      # Unigram Weight

BETA = 1.0       # Bigram Weight

LAMBDA = 0.01    # Redundancy Penalty

# ==========================================================
# SCORE FUNCTION
# ==========================================================

def score_sentence(sentence):

    # ----------------------------------------
    # Remove spaces
    # ----------------------------------------

    chars = list(
        sentence.replace(" ", "").strip()
    )

    if len(chars) == 0:
        return 0

    # ----------------------------------------
    # Character Unigram
    # ----------------------------------------

    unigram = Counter(chars)

    # ----------------------------------------
    # Character Bigram
    # ----------------------------------------

    bigram = Counter()

    for i in range(len(chars)-1):

        pair = (chars[i], chars[i+1])

        bigram[pair] += 1

    # ----------------------------------------
    # Character Unigram Gain
    # ----------------------------------------

    unigram_gain = 0

    for ch in unigram:

        if covered_unigrams[ch] == 0:

            unigram_gain += 1

    # ----------------------------------------
    # Character Bigram Gain
    # ----------------------------------------

    bigram_gain = 0

    for bg in bigram:

        if covered_bigrams[bg] == 0:

            bigram_gain += 1

    # ----------------------------------------
    # Redundancy Penalty
    # ----------------------------------------

    redundancy = 0

    for ch in unigram:

        redundancy += covered_unigrams[ch]

    # ----------------------------------------
    # Final Score
    # ----------------------------------------

    score = (
        ALPHA * unigram_gain
        +
        BETA * bigram_gain
        -
        LAMBDA * redundancy
    )

    return score
```

```python
# ============================================================
# PART A: INFRASTRUCTURE + FEATURE ENGINE
# For Urdu IPA merged corpus
# Builds:
# 1. sentence byte offsets
# 2. unigram vocabulary
# 3. bigram vocabulary
# 4. disk-based feature cache
# ============================================================

import os
import json
import pickle
import struct
from array import array
from collections import Counter
from tqdm import tqdm
```

```python
# ============================================================
# A1: CONFIG
# ============================================================

FULL_CORPUS_PATH = "/content/drive/MyDrive/Thesis 2/ipa_merged.txt"

BASE_OUTPUT_DIR = "/content/drive/MyDrive/Thesis 2/urdu_greedy_outputs"
FEATURE_DIR = os.path.join(BASE_OUTPUT_DIR, "feature_cache")

os.makedirs(BASE_OUTPUT_DIR, exist_ok=True)
os.makedirs(FEATURE_DIR, exist_ok=True)

# Testing:
# SAMPLE_LIMIT = 50000
# SAMPLE_LIMIT = 100000

# Full corpus:
SAMPLE_LIMIT = None

IGNORE_SPACES = True

OFFSETS_PATH = os.path.join(FEATURE_DIR, "sentence_offsets.bin")
FEATURE_OFFSETS_PATH = os.path.join(FEATURE_DIR, "feature_offsets.bin")
FEATURE_DATA_PATH = os.path.join(FEATURE_DIR, "feature_data.bin")

UNIGRAM_VOCAB_PATH = os.path.join(FEATURE_DIR, "unigram_vocab.pkl")
BIGRAM_VOCAB_PATH = os.path.join(FEATURE_DIR, "bigram_vocab.pkl")

FULL_UNIGRAM_COUNTS_PATH = os.path.join(FEATURE_DIR, "full_unigram_counts.pkl")
FULL_BIGRAM_COUNTS_PATH = os.path.join(FEATURE_DIR, "full_bigram_counts.pkl")

METADATA_PATH = os.path.join(FEATURE_DIR, "partA_metadata.json")
```

```python
# ============================================================
# A2: BASIC TEXT + FEATURE FUNCTIONS
# ============================================================

def clean_sentence(line):
    line = line.strip()

    if IGNORE_SPACES:
        line = line.replace(" ", "")

    return line


def extract_unigrams_bigrams(sentence):
    chars = list(sentence)

    unigrams = set(chars)

    bigrams = set()
    for i in range(len(chars) - 1):
        bigrams.add(chars[i] + chars[i + 1])

    return unigrams, bigrams


def update_full_counts(sentence, unigram_counter, bigram_counter):
    chars = list(sentence)

    unigram_counter.update(chars)

    for i in range(len(chars) - 1):
        bigram_counter[chars[i] + chars[i + 1]] += 1
```

```python
# ============================================================
# A3: PASS 1
# Build:
# - sentence offsets
# - unigram vocabulary
# - bigram vocabulary
# - full unigram counts
# - full bigram counts
# ============================================================

def build_offsets_vocab_and_counts(file_path, sample_limit=None):
    sentence_offsets = array("Q")

    unigram_to_id = {}
    bigram_to_id = {}

    full_unigram_counts = Counter()
    full_bigram_counts = Counter()

    sentence_count = 0

    with open(file_path, "rb") as f:

        while True:
            byte_offset = f.tell()
            raw_line = f.readline()

            if not raw_line:
                break

            line = raw_line.decode("utf-8", errors="ignore")
            sentence = clean_sentence(line)

            if not sentence:
                continue

            sentence_offsets.append(byte_offset)

            unigrams, bigrams = extract_unigrams_bigrams(sentence)

            for u in unigrams:
                if u not in unigram_to_id:
                    unigram_to_id[u] = len(unigram_to_id)

            for b in bigrams:
                if b not in bigram_to_id:
                    bigram_to_id[b] = len(bigram_to_id)

            update_full_counts(
                sentence,
                full_unigram_counts,
                full_bigram_counts
            )

            sentence_count += 1

            if sample_limit is not None and sentence_count >= sample_limit:
                break

    print("Pass 1 completed.")
    print("Total usable sentences:", sentence_count)
    print("Unique unigrams:", len(unigram_to_id))
    print("Unique bigrams:", len(bigram_to_id))

    return (
        sentence_offsets,
        unigram_to_id,
        bigram_to_id,
        full_unigram_counts,
        full_bigram_counts
    )
```

```python
# ============================================================
# A4: SAVE / LOAD OFFSETS AND OBJECTS
# ============================================================

def save_array_uint64(path, arr):
    with open(path, "wb") as f:
        arr.tofile(f)


def load_array_uint64(path):
    arr = array("Q")
    with open(path, "rb") as f:
        arr.fromfile(f, os.path.getsize(path) // arr.itemsize)
    return arr


def save_pickle(path, obj):
    with open(path, "wb") as f:
        pickle.dump(obj, f, protocol=pickle.HIGHEST_PROTOCOL)


def load_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)
```

```python
# ============================================================
# A5: PASS 2
# Build disk-based feature cache
#
# For every sentence, store:
# [num_unigrams][num_bigrams][unigram_ids][bigram_ids]
#
# Format:
# - num_unigrams: unsigned short, 2 bytes
# - num_bigrams: unsigned short, 2 bytes
# - each feature id: unsigned short, 2 bytes
#
# This is memory-safe because features are saved to disk.
# ============================================================

def sentence_to_feature_ids(sentence, unigram_to_id, bigram_to_id):
    unigrams, bigrams = extract_unigrams_bigrams(sentence)

    unigram_ids = sorted(unigram_to_id[u] for u in unigrams)
    bigram_ids = sorted(bigram_to_id[b] for b in bigrams)

    return unigram_ids, bigram_ids


def build_feature_cache(
    file_path,
    sentence_offsets,
    unigram_to_id,
    bigram_to_id,
    feature_data_path,
    feature_offsets_path
):
    feature_offsets = array("Q")

    with open(file_path, "rb") as f_in, open(feature_data_path, "wb") as f_out:

        for idx in tqdm(range(len(sentence_offsets)), desc="Building feature cache"):
            feature_offsets.append(f_out.tell())

            f_in.seek(sentence_offsets[idx])
            line = f_in.readline().decode("utf-8", errors="ignore")
            sentence = clean_sentence(line)

            unigram_ids, bigram_ids = sentence_to_feature_ids(
                sentence,
                unigram_to_id,
                bigram_to_id
            )

            if len(unigram_ids) > 65535 or len(bigram_ids) > 65535:
                raise ValueError("Too many features in one sentence for uint16 format.")

            record = struct.pack("<HH", len(unigram_ids), len(bigram_ids))

            if unigram_ids:
                record += array("H", unigram_ids).tobytes()

            if bigram_ids:
                record += array("H", bigram_ids).tobytes()

            f_out.write(record)

    save_array_uint64(feature_offsets_path, feature_offsets)

    print("Feature cache completed.")
    print("Feature records:", len(feature_offsets))
    print("Feature data file:", feature_data_path)
```

```python
# ============================================================
# A6: READ FEATURES FROM DISK CACHE
# This function will be used in Part B CELF Lazy Greedy.
# ============================================================

def read_feature_ids(feature_file_handle, feature_offsets, sentence_idx):
    feature_file_handle.seek(feature_offsets[sentence_idx])

    header = feature_file_handle.read(4)
    num_unigrams, num_bigrams = struct.unpack("<HH", header)

    unigram_bytes = feature_file_handle.read(num_unigrams * 2)
    bigram_bytes = feature_file_handle.read(num_bigrams * 2)

    unigram_ids = array("H")
    bigram_ids = array("H")

    if num_unigrams > 0:
        unigram_ids.frombytes(unigram_bytes)

    if num_bigrams > 0:
        bigram_ids.frombytes(bigram_bytes)

    return unigram_ids, bigram_ids
```

Part B

```python
# ============================================================
# PART B: TRUE CELF-STYLE LAZY GREEDY SELECTION
# Uses Part A disk-based feature cache
#
# Output:
# selected_subset_5percent.txt
# selected_subset_10percent.txt
# selected_subset_20percent.txt
# ============================================================

import os
import json
import pickle
import heapq
import struct
import gc
from array import array
from tqdm import tqdm
```

```python
# ============================================================
# B1: CONFIG
# ============================================================

BASE_OUTPUT_DIR = "/content/drive/MyDrive/Thesis 2/urdu_greedy_outputs"
FEATURE_DIR = os.path.join(BASE_OUTPUT_DIR, "feature_cache")

METADATA_PATH = os.path.join(FEATURE_DIR, "partA_metadata.json")

CHECKPOINT_PATH = os.path.join(
    BASE_OUTPUT_DIR,
    "PARTB_TRUE_CELF_checkpoint.pkl"
)

TARGET_PERCENTAGES = [0.001, 0.01, 0.05, 0.10, 0.15,0.20]

ALPHA = 1.0
BETA = 1.0
LAMBDA = 0.5

CHECKPOINT_EVERY = 1000
PROGRESS_EVERY = 500
```

```python
# ============================================================
# B2: LOAD PART A METADATA AND FILE PATHS
# ============================================================

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

FULL_CORPUS_PATH = metadata["full_corpus_path"]

OFFSETS_PATH = metadata["offsets_path"]
FEATURE_OFFSETS_PATH = metadata["feature_offsets_path"]
FEATURE_DATA_PATH = metadata["feature_data_path"]

UNIGRAM_VOCAB_PATH = metadata["unigram_vocab_path"]
BIGRAM_VOCAB_PATH = metadata["bigram_vocab_path"]

TOTAL_SENTENCES = metadata["total_sentences"]
UNIGRAM_TYPES = metadata["unigram_types"]
BIGRAM_TYPES = metadata["bigram_types"]

print("Loaded Part A metadata")
print("Total sentences:", TOTAL_SENTENCES)
print("Unigram types:", UNIGRAM_TYPES)
print("Bigram types:", BIGRAM_TYPES)
```

```python
# ============================================================
# B3: LOAD ARRAYS AND VOCABULARIES
# ============================================================

def load_array_uint64(path):
    arr = array("Q")
    with open(path, "rb") as f:
        arr.fromfile(f, os.path.getsize(path) // arr.itemsize)
    return arr


def load_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)


sentence_offsets = load_array_uint64(OFFSETS_PATH)
feature_offsets = load_array_uint64(FEATURE_OFFSETS_PATH)

unigram_to_id = load_pickle(UNIGRAM_VOCAB_PATH)
bigram_to_id = load_pickle(BIGRAM_VOCAB_PATH)

print("Loaded offsets and vocabularies")
print("Sentence offsets:", len(sentence_offsets))
print("Feature offsets:", len(feature_offsets))
```

```python
# ============================================================
# B4: READ FEATURE IDS FROM DISK CACHE
# ============================================================

def read_feature_ids(feature_file_handle, feature_offsets, sentence_idx):
    feature_file_handle.seek(feature_offsets[sentence_idx])

    header = feature_file_handle.read(4)

    if len(header) < 4:
        raise ValueError(f"Invalid feature record at sentence index {sentence_idx}")

    num_unigrams, num_bigrams = struct.unpack("<HH", header)

    unigram_bytes = feature_file_handle.read(num_unigrams * 2)
    bigram_bytes = feature_file_handle.read(num_bigrams * 2)

    unigram_ids = array("H")
    bigram_ids = array("H")

    if num_unigrams > 0:
        unigram_ids.frombytes(unigram_bytes)

    if num_bigrams > 0:
        bigram_ids.frombytes(bigram_bytes)

    return unigram_ids, bigram_ids
```

```python
# ============================================================
# B5: KL-AWARE DEFICIT SCORE FUNCTION
# ============================================================

import pickle
import math

FULL_UNIGRAM_COUNTS_PATH = metadata["full_unigram_counts_path"]
FULL_BIGRAM_COUNTS_PATH = metadata["full_bigram_counts_path"]

with open(FULL_UNIGRAM_COUNTS_PATH, "rb") as f:
    full_unigram_counts = pickle.load(f)

with open(FULL_BIGRAM_COUNTS_PATH, "rb") as f:
    full_bigram_counts = pickle.load(f)

full_unigram_total = sum(full_unigram_counts.values())
full_bigram_total = sum(full_bigram_counts.values())

full_unigram_probs = [0.0] * UNIGRAM_TYPES
full_bigram_probs = [0.0] * BIGRAM_TYPES

for token, idx in unigram_to_id.items():
    full_unigram_probs[idx] = full_unigram_counts.get(token, 0) / full_unigram_total

for token, idx in bigram_to_id.items():
    full_bigram_probs[idx] = full_bigram_counts.get(token, 0) / full_bigram_total


GAMMA = 1.0
LENGTH_NORM_POWER = 0.5


def compute_feature_score(
    unigram_ids,
    bigram_ids,
    covered_unigrams,
    covered_bigrams
):
    selected_unigram_total = sum(covered_unigrams)
    selected_bigram_total = sum(covered_bigrams)

    unigram_score = 0.0
    bigram_score = 0.0
    penalty = 0.0

    for uid in unigram_ids:
        p_full = full_unigram_probs[uid]
        p_selected = 0.0 if selected_unigram_total == 0 else covered_unigrams[uid] / selected_unigram_total

        deficit = max(0.0, p_full - p_selected)
        excess = max(0.0, p_selected - p_full)

        unigram_score += deficit
        penalty += excess

    for bid in bigram_ids:
        p_full = full_bigram_probs[bid]
        p_selected = 0.0 if selected_bigram_total == 0 else covered_bigrams[bid] / selected_bigram_total

        deficit = max(0.0, p_full - p_selected)
        excess = max(0.0, p_selected - p_full)

        bigram_score += deficit
        penalty += excess

    score = (ALPHA * unigram_score) + (BETA * bigram_score) - (GAMMA * penalty)

    length = len(unigram_ids) + len(bigram_ids)
    if length > 0:
        score = score / (length ** LENGTH_NORM_POWER)

    return score
```

```python
# ============================================================
# B6: CHECKPOINT FUNCTIONS
# ============================================================

def save_checkpoint(
    selected_indices,
    selected_mask,
    covered_unigrams,
    covered_bigrams,
    saved_percentages
):
    state = {
        "selected_indices": selected_indices,
        "selected_mask": selected_mask,
        "covered_unigrams": covered_unigrams,
        "covered_bigrams": covered_bigrams,
        "saved_percentages": saved_percentages,
        "target_percentages": TARGET_PERCENTAGES
    }

    with open(CHECKPOINT_PATH, "wb") as f:
        pickle.dump(state, f, protocol=pickle.HIGHEST_PROTOCOL)

    print("Checkpoint saved:", len(selected_indices))


def load_checkpoint():
    if os.path.exists(CHECKPOINT_PATH):
        with open(CHECKPOINT_PATH, "rb") as f:
            state = pickle.load(f)

        same_size = len(state.get("selected_mask", [])) == TOTAL_SENTENCES
        same_targets = state.get("target_percentages") == TARGET_PERCENTAGES

        if same_size and same_targets:
            print("Checkpoint loaded")
            print("Already selected:", len(state["selected_indices"]))
            return state

        print("Checkpoint ignored because corpus size or target percentages changed.")

    return {
        "selected_indices": [],
        "selected_mask": bytearray(TOTAL_SENTENCES),
        "covered_unigrams": [0] * UNIGRAM_TYPES,
        "covered_bigrams": [0] * BIGRAM_TYPES,
        "saved_percentages": set()
    }
```

```python
# ============================================================
# B7: BUILD PRIORITY QUEUE
#
# This builds cached marginal score for every unselected sentence.
# Python heapq is min-heap, so negative score is stored.
# ============================================================

def build_priority_queue(
    feature_data_path,
    feature_offsets,
    selected_mask,
    covered_unigrams,
    covered_bigrams
):
    heap = []

    with open(feature_data_path, "rb") as f_feat:

        for idx in tqdm(range(TOTAL_SENTENCES), desc="Building CELF priority queue"):
            if selected_mask[idx]:
                continue

            unigram_ids, bigram_ids = read_feature_ids(
                f_feat,
                feature_offsets,
                idx
            )

            score = compute_feature_score(
                unigram_ids,
                bigram_ids,
                covered_unigrams,
                covered_bigrams
            )

            heapq.heappush(heap, (-score, idx))

    print("Priority queue size:", len(heap))
    return heap
```

```python
# ============================================================
# B8: READ ORIGINAL SENTENCE AND SAVE SUBSET
# ============================================================

def read_original_sentence(corpus_file_handle, sentence_offsets, sentence_idx):
    corpus_file_handle.seek(sentence_offsets[sentence_idx])
    line = corpus_file_handle.readline().decode("utf-8", errors="ignore")
    return line.strip()


def percentage_label(p):
    if p == 0.001:
        return "0_1percent"
    elif p == 0.01:
        return "1percent"
    else:
        return f"{int(p * 100)}percent"


def save_subset_file(percentage, selected_indices):
    label = percentage_label(percentage)

    output_path = os.path.join(
        BASE_OUTPUT_DIR,
        f"selected_subset_{label}.txt"
    )

    with open(FULL_CORPUS_PATH, "rb") as f_corpus, open(output_path, "w", encoding="utf-8") as out:
        for idx in tqdm(sorted(selected_indices), desc=f"Saving {label} subset"):
            sentence = read_original_sentence(
                f_corpus,
                sentence_offsets,
                idx
            )
            if sentence:
                out.write(sentence + "\n")

    print("Saved:", output_path)
```

```python
# ============================================================
# B9: UPDATE FEATURE COUNTS AFTER SELECTION
# ============================================================

def update_coverage(
    unigram_ids,
    bigram_ids,
    covered_unigrams,
    covered_bigrams
):
    for uid in unigram_ids:
        covered_unigrams[uid] += 1

    for bid in bigram_ids:
        covered_bigrams[bid] += 1
```

```python
# ============================================================
# B10: TRUE CELF LAZY GREEDY SELECTION
#
# Strict CELF condition:
# 1. Pop highest cached-score candidate.
# 2. Recompute its score using current coverage.
# 3. Compare updated score with next best cached score in heap.
# 4. Select only if updated score is still >= next best cached score.
# 5. Otherwise push it back with updated score.
# ============================================================

def true_celf_lazy_greedy_selection():
    target_sizes = {
        p: int(TOTAL_SENTENCES * p)
        for p in TARGET_PERCENTAGES
    }

    max_target = max(target_sizes.values())

    print("Target subset sizes:")
    for p, size in target_sizes.items():
        print(f"{int(p * 100)}% = {size} sentences")

    state = load_checkpoint()

    selected_indices = state["selected_indices"]
    selected_mask = state["selected_mask"]
    covered_unigrams = state["covered_unigrams"]
    covered_bigrams = state["covered_bigrams"]
    saved_percentages = state["saved_percentages"]

    heap = build_priority_queue(
        FEATURE_DATA_PATH,
        feature_offsets,
        selected_mask,
        covered_unigrams,
        covered_bigrams
    )

    with open(FEATURE_DATA_PATH, "rb") as f_feat:

        while len(selected_indices) < max_target and heap:

            cached_negative_score, idx = heapq.heappop(heap)

            if selected_mask[idx]:
                continue

            unigram_ids, bigram_ids = read_feature_ids(
                f_feat,
                feature_offsets,
                idx
            )

            updated_score = compute_feature_score(
                unigram_ids,
                bigram_ids,
                covered_unigrams,
                covered_bigrams
            )

            updated_negative_score = -updated_score

            # Strict CELF comparison with next best cached candidate
            if not heap or updated_negative_score <= heap[0][0]:

                selected_indices.append(idx)
                selected_mask[idx] = 1

                update_coverage(
                    unigram_ids,
                    bigram_ids,
                    covered_unigrams,
                    covered_bigrams
                )

                current_count = len(selected_indices)

                if current_count % PROGRESS_EVERY == 0:
                    print(
                        f"Selected {current_count}/{max_target} | "
                        f"Covered unigrams: {sum(covered_unigrams)}/{UNIGRAM_TYPES} | "
                        f"Covered bigrams: {sum(covered_bigrams)}/{BIGRAM_TYPES}"
                    )

                for p, target in target_sizes.items():
                    if current_count >= target and p not in saved_percentages:
                        save_subset_file(
                            p,
                            selected_indices
                        )
                        saved_percentages.add(p)

                if current_count % CHECKPOINT_EVERY == 0:
                    save_checkpoint(
                        selected_indices,
                        selected_mask,
                        covered_unigrams,
                        covered_bigrams,
                        saved_percentages
                    )
                    gc.collect()

            else:
                heapq.heappush(heap, (updated_negative_score, idx))

    save_checkpoint(
        selected_indices,
        selected_mask,
        covered_unigrams,
        covered_bigrams,
        saved_percentages
    )

    print("Part B completed.")
    print("Total selected:", len(selected_indices))
    print("Covered unigrams:", sum(covered_unigrams), "/", UNIGRAM_TYPES)
    print("Covered bigrams:", sum(covered_bigrams), "/", BIGRAM_TYPES)

    return selected_indices, covered_unigrams, covered_bigrams
```

```python
# ============================================================
# B11: RUN PART B
# ============================================================

selected_indices, covered_unigrams, covered_bigrams = true_celf_lazy_greedy_selection()
```

```python
# ============================================================
# PART C: CHECKPOINT, RESUME STATUS, OUTPUT VALIDATION
# For Part A + Part B pipeline
# ============================================================

import os
import json
import pickle
from array import array
from tqdm import tqdm
```

```python
# ============================================================
# C2: LOAD METADATA
# ============================================================

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

TOTAL_SENTENCES = metadata["total_sentences"]
FULL_CORPUS_PATH = metadata["full_corpus_path"]

TARGET_SIZES = {
    p: int(TOTAL_SENTENCES * p)
    for p in SUBSET_PATHS.keys()
}

print("Part C loaded metadata.")
print("Full corpus path:", FULL_CORPUS_PATH)
print("Total sentences:", TOTAL_SENTENCES)
print("Target sizes:", TARGET_SIZES)
```

```python
# ============================================================
# C3: COUNT SENTENCES IN A FILE
# ============================================================

def count_nonempty_lines(file_path):
    count = 0

    if not os.path.exists(file_path):
        return 0

    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            if line.strip():
                count += 1

    return count
```

```python
# ============================================================
# C4: CHECK CHECKPOINT STATUS
# ============================================================

def check_checkpoint_status():
    if not os.path.exists(CHECKPOINT_PATH):
        print("No checkpoint found.")
        return None

    with open(CHECKPOINT_PATH, "rb") as f:
        state = pickle.load(f)

    selected_count = len(state.get("selected_indices", []))

    print("Checkpoint found.")
    print("Selected sentences in checkpoint:", selected_count)
    print("Selected mask length:", len(state.get("selected_mask", [])))
    print("Covered unigrams:", sum(state.get("covered_unigrams", [])))
    print("Covered bigrams:", sum(state.get("covered_bigrams", [])))
    print("Saved percentages:", state.get("saved_percentages", set()))

    if len(state.get("selected_mask", [])) != TOTAL_SENTENCES:
        print("WARNING: checkpoint sentence count does not match current corpus.")

    return state


checkpoint_state = check_checkpoint_status()
```

```python
# ============================================================
# C5: VALIDATE SUBSET FILES
# ============================================================

def validate_subset_files():
    validation_rows = []

    for percentage, path in SUBSET_PATHS.items():
        expected = TARGET_SIZES[percentage]
        actual = count_nonempty_lines(path)

        status = "OK" if actual == expected else "MISSING_OR_INCOMPLETE"

        row = {
            "percentage": int(percentage * 100),
            "expected_sentences": expected,
            "actual_sentences": actual,
            "status": status,
            "path": path
        }

        validation_rows.append(row)

    return validation_rows


validation_rows = validate_subset_files()

for row in validation_rows:
    print(row)
```

Part D

```python
# ============================================================
# PART D: FINAL EVALUATION
# Metrics:
# 1. Type Coverage
# 2. Token Probability Coverage
# 3. KL Divergence
#
# Applied on:
# Character-level Unigrams
# Character-level Bigrams
# ============================================================

import os
import csv
import math
import json
from collections import Counter
from tqdm import tqdm
```

```python
# ============================================================
# D1: CONFIG
# ============================================================

BASE_OUTPUT_DIR = "/content/drive/MyDrive/Thesis 2/urdu_greedy_outputs"
FEATURE_DIR = os.path.join(BASE_OUTPUT_DIR, "feature_cache")

METADATA_PATH = os.path.join(FEATURE_DIR, "partA_metadata.json")

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

FULL_CORPUS_PATH = metadata["full_corpus_path"]

FULL_UNIGRAM_COUNTS_PATH = metadata["full_unigram_counts_path"]
FULL_BIGRAM_COUNTS_PATH = metadata["full_bigram_counts_path"]

SUBSET_FILES = {
    "greedy_0_1_percent": os.path.join(BASE_OUTPUT_DIR, "selected_subset_0_1percent.txt"),
    "greedy_1_percent": os.path.join(BASE_OUTPUT_DIR, "selected_subset_1percent.txt"),
    "greedy_5_percent": os.path.join(BASE_OUTPUT_DIR, "selected_subset_5percent.txt"),
    "greedy_10_percent": os.path.join(BASE_OUTPUT_DIR, "selected_subset_10percent.txt"),
    "greedy_15_percent": os.path.join(BASE_OUTPUT_DIR, "selected_subset_15percent.txt"),
    "greedy_20_percent": os.path.join(BASE_OUTPUT_DIR, "selected_subset_20percent.txt"),

}

EVALUATION_OUTPUT_CSV = os.path.join(
    BASE_OUTPUT_DIR,
    "PARTD_algorithm1_updated_evaluation_results.csv"
)

SMOOTHING_ALPHA = 1e-12
IGNORE_SPACES = True
```

```python
# ============================================================
# D2: LOAD FULL CORPUS COUNTS FROM PART A
# ============================================================

import pickle

def load_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)


full_unigram_counts = load_pickle(FULL_UNIGRAM_COUNTS_PATH)
full_bigram_counts = load_pickle(FULL_BIGRAM_COUNTS_PATH)

print("Loaded full corpus counts from Part A.")
print("Full unigram types:", len(full_unigram_counts))
print("Full bigram types:", len(full_bigram_counts))
print("Full unigram tokens:", sum(full_unigram_counts.values()))
print("Full bigram tokens:", sum(full_bigram_counts.values()))
```

```python
# ============================================================
# D3: COUNT SUBSET UNIGRAMS AND BIGRAMS
# ============================================================

def clean_eval_sentence(line):
    line = line.strip()
    if IGNORE_SPACES:
        line = line.replace(" ", "")
    return line


def update_ngram_counts(sentence, unigram_counter, bigram_counter):
    chars = list(sentence)

    unigram_counter.update(chars)

    for i in range(len(chars) - 1):
        bigram_counter[chars[i] + chars[i + 1]] += 1


def compute_subset_counts(file_path):
    unigram_counter = Counter()
    bigram_counter = Counter()
    sentence_count = 0

    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        for line in tqdm(f, desc=f"Counting {os.path.basename(file_path)}"):
            sentence = clean_eval_sentence(line)

            if not sentence:
                continue

            update_ngram_counts(sentence, unigram_counter, bigram_counter)
            sentence_count += 1

    return unigram_counter, bigram_counter, sentence_count
```

```python
# ============================================================
# D4: EVALUATION METRICS
# ============================================================

def type_coverage(full_counter, subset_counter):
    V_full = set(full_counter.keys())
    V_subset = set(subset_counter.keys())

    if len(V_full) == 0:
        return 0.0

    return len(V_subset & V_full) / len(V_full)


def token_probability_coverage(full_counter, subset_counter):
    total_full_tokens = sum(full_counter.values())

    if total_full_tokens == 0:
        return 0.0

    V_subset = set(subset_counter.keys())

    probability_mass = 0.0

    for token in V_subset:
        probability_mass += full_counter.get(token, 0) / total_full_tokens

    return probability_mass


def kl_divergence(full_counter, subset_counter, alpha=SMOOTHING_ALPHA):
    vocab = set(full_counter.keys()) | set(subset_counter.keys())

    full_total = sum(full_counter.values())
    subset_total = sum(subset_counter.values())
    vocab_size = len(vocab)

    if full_total == 0 or subset_total == 0:
        return float("inf")

    kl = 0.0

    for token in vocab:
        p_full = (full_counter.get(token, 0) + alpha) / (
            full_total + alpha * vocab_size
        )

        p_subset = (subset_counter.get(token, 0) + alpha) / (
            subset_total + alpha * vocab_size
        )

        kl += p_full * math.log(p_full / p_subset)

    return kl
```

```python
# ============================================================
# D5: EVALUATE ALL GREEDY SUBSETS
# ============================================================

def evaluate_subset(subset_name, subset_path):
    subset_unigrams, subset_bigrams, sentence_count = compute_subset_counts(
        subset_path
    )

    result = {
        "subset_name": subset_name,
        "sentences": sentence_count,

        "unigram_type_coverage": type_coverage(
            full_unigram_counts,
            subset_unigrams
        ),

        "bigram_type_coverage": type_coverage(
            full_bigram_counts,
            subset_bigrams
        ),

        "unigram_token_probability_coverage": token_probability_coverage(
            full_unigram_counts,
            subset_unigrams
        ),

        "bigram_token_probability_coverage": token_probability_coverage(
            full_bigram_counts,
            subset_bigrams
        ),

        "unigram_kl_divergence": kl_divergence(
            full_unigram_counts,
            subset_unigrams
        ),

        "bigram_kl_divergence": kl_divergence(
            full_bigram_counts,
            subset_bigrams
        )
    }

    return result


results = []

for subset_name, subset_path in SUBSET_FILES.items():
    if not os.path.exists(subset_path):
        print("Missing subset file, skipped:", subset_path)
        continue

    print("\nEvaluating:", subset_name)

    result = evaluate_subset(subset_name, subset_path)
    results.append(result)

    print(result)
```

```python
# ============================================================
# RANDOM BASELINE GENERATION + EVALUATION
# For 0.1%, 1%, 5%, 10%, 15% 20%
# ============================================================

import os
import random
import csv
import pandas as pd
from tqdm import tqdm

RANDOM_SEED = 42

RANDOM_SUBSET_FILES = {
    "random_0_1_percent": os.path.join(BASE_OUTPUT_DIR, "random_subset_0_1percent.txt"),
    "random_1_percent": os.path.join(BASE_OUTPUT_DIR, "random_subset_1percent.txt"),
    "random_5_percent": os.path.join(BASE_OUTPUT_DIR, "random_subset_5percent.txt"),
    "random_10_percent": os.path.join(BASE_OUTPUT_DIR, "random_subset_10percent.txt"),
    "random_15_percent": os.path.join(BASE_OUTPUT_DIR, "random_subset_15percent.txt"),
    "random_20_percent": os.path.join(BASE_OUTPUT_DIR, "random_subset_20percent.txt"),
}

RANDOM_TARGETS = {
    "random_0_1_percent": 5696,
    "random_1_percent": 56968,
    "random_5_percent": 284843,
    "random_10_percent": 569687,
    "random_15_percent": 854530,
    "random_20_percent": 1139375,
}
```

```python
 # ============================================================
# CREATE RANDOM SUBSET USING RESERVOIR SAMPLING
# Memory-safe for 5.7M corpus
# ============================================================

def create_random_subset(full_corpus_path, output_path, target_size, seed=42):
    random.seed(seed)

    reservoir = []
    seen = 0

    with open(full_corpus_path, "r", encoding="utf-8", errors="ignore") as f:
        for line in tqdm(f, desc=f"Creating random subset {target_size}"):
            sentence = line.strip()

            if not sentence:
                continue

            seen += 1

            if len(reservoir) < target_size:
                reservoir.append(sentence)
            else:
                j = random.randint(0, seen - 1)
                if j < target_size:
                    reservoir[j] = sentence

    with open(output_path, "w", encoding="utf-8") as out:
        for sentence in reservoir:
            out.write(sentence + "\n")

    print("Saved random subset:", output_path)
```

```python
# ============================================================
# GENERATE RANDOM SUBSETS
# ============================================================

for name, output_path in RANDOM_SUBSET_FILES.items():
    target_size = RANDOM_TARGETS[name]

    if os.path.exists(output_path):
        print("Already exists, skipped:", output_path)
    else:
        create_random_subset(
            FULL_CORPUS_PATH,
            output_path,
            target_size,
            seed=RANDOM_SEED
        )
```

```python
# ============================================================
# EVALUATE RANDOM SUBSETS
# Uses same evaluate_subset() function from Part D
# ============================================================

random_results = []

for subset_name, subset_path in RANDOM_SUBSET_FILES.items():
    if not os.path.exists(subset_path):
        print("Missing random subset:", subset_path)
        continue

    print("\nEvaluating:", subset_name)

    result = evaluate_subset(subset_name, subset_path)
    random_results.append(result)

    print(result)
```

```python
# ============================================================
# FINAL KL COMPARISON TABLE
# ============================================================

def extract_size_label(name):
    if "0_1" in name:
        return "0.1%"
    elif "1_percent" in name:
        return "1%"
    elif "5_percent" in name:
        return "5%"
    elif "10_percent" in name:
        return "10%"
    elif "15_percent" in name:
        return "15%"
    elif "20_percent" in name:
        return "20%"
    return name


comparison_rows = []

for size_label in ["0.1%", "1%", "5%", "10%", "15%","20%"]:
    greedy_row = combined_df[
        (combined_df["subset_name"].str.contains("greedy")) &
        (combined_df["subset_name"].apply(extract_size_label) == size_label)
    ]

    random_row = combined_df[
        (combined_df["subset_name"].str.contains("random")) &
        (combined_df["subset_name"].apply(extract_size_label) == size_label)
    ]

    if len(greedy_row) == 0 or len(random_row) == 0:
        continue

    greedy_row = greedy_row.iloc[0]
    random_row = random_row.iloc[0]

    comparison_rows.append({
        "subset_size": size_label,
        "greedy_unigram_kl": greedy_row["unigram_kl_divergence"],
        "random_unigram_kl": random_row["unigram_kl_divergence"],
        "greedy_bigram_kl": greedy_row["bigram_kl_divergence"],
        "random_bigram_kl": random_row["bigram_kl_divergence"],
        "unigram_better": "Greedy" if greedy_row["unigram_kl_divergence"] < random_row["unigram_kl_divergence"] else "Random",
        "bigram_better": "Greedy" if greedy_row["bigram_kl_divergence"] < random_row["bigram_kl_divergence"] else "Random",
    })

comparison_df = pd.DataFrame(comparison_rows)

COMPARISON_CSV = os.path.join(
    BASE_OUTPUT_DIR,
    "FINAL_GREEDY_vs_RANDOM_KL_TABLE.csv"
)

comparison_df.to_csv(COMPARISON_CSV, index=False)

comparison_df
```
